```python
import cv2
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow

img = cv2.imread("/content/drive/MyDrive/PCVK2026/ktp1.jpg")
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.show()

img_gray = cv2.imread("/content/drive/MyDrive/PCVK2026/ktp1.jpg", cv2.IMREAD_GRAYSCALE)

plt.subplot(1,2,1)
plt.imshow(img_gray)
plt.title("Tanpa Cmap")

plt.subplot(1,2,2)
plt.imshow(img_gray, cmap="gray")
plt.title("Dengan Cmap")

plt.show()

imgRGB=cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
imgRGBsmall=cv2.resize(imgRGB,(400,250))
plt.imshow(imgRGBsmall)

imgBalik=cv2.flip(imgRGBsmall,1)

fig=plt.figure(figsize=(10,10))
ax=fig.add_subplot(111)
ax.imshow(imgBalik)
plt.imshow(imgBalik)
```
